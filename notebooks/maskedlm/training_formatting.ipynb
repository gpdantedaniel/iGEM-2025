## Models and datasets

```python
model_names = {
    'facebook/esm2_t30_150M_UR50D': 'ESM2-150M',
    'facebook/esm2_t33_650M_UR50D': 'ESM2-650M',
    'facebook/esm1v_t33_650M_UR90S_1': 'ESM1V-650M'
}

dataset_names = {
    'dantedgp/fnbp-70K-tokenized': 'FNBP-70K',
    'dantedgp/mscramm-100K-tokenized': 'MSCRAMM-100K',
    'dantedgp/mscramm-227K-tokenized': 'MSCRAMM-227K',
    'dantedgp/cwa-500K-tokenized': 'CWA-500K',
    'dantedgp/cwa-1M-tokenized': 'CWA-1M'
}

training_jobs = {
    'facebook/esm2_t30_150M_UR50D': [
        'dantedgp/fnbp-70K-tokenized',
        'dantedgp/mscramm-100K-tokenized',
        'dantedgp/mscramm-227K-tokenized'
    ],
    'facebook/esm1v_t33_650M_UR90S_1': [
        'dantedgp/mscramm-100K-tokenized',
        'dantedgp/mscramm-227K-tokenized',
        'dantedgp/cwa-500K-tokenized',
        'dantedgp/cwa-1M-tokenized'
    ]
}
```

```python
import os

def format_script(template, dataset, base_model, output_model):
    script_content = template
    script_content = script_content.replace('{DATASET}', dataset)
    script_content = script_content.replace('{BASE_MODEL}', base_model)
    script_content = script_content.replace('{OUTPUT_MODEL}', output_model)

    return script_content

template_path = 'training_template.py'

with open(template_path, 'r') as file:
    template = file.read()

for base_model in training_jobs.keys():
    for dataset in training_jobs[base_model]:
        output_model = f'SpySeer-{model_names[base_model]}-{dataset_names[dataset]}'
        script_content = format_script(template, dataset, base_model, output_model)
        output_path = f'training_scripts/train-{output_model}.py'
        
        with open(output_path, 'w') as file:
            file.write(script_content)
```
